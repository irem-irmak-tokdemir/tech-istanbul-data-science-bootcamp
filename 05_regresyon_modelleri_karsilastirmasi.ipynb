```python
from sklearn.datasets import load_diabetes
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.cross_decomposition import PLSRegression
from sklearn.decomposition import PCA
from sklearn.metrics import mean_squared_error, r2_score
diabetes = load_diabetes(scaled=False)
```

```python
x= diabetes.data
y = diabetes.target
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
ölçek= StandardScaler()
x_ölçekli=ölçek.fit_transform(x_train)
x_ölçekli_test=ölçek.transform(x_test)
```

# Doğrusal regrasyon

```python
model_dogrusal=LinearRegression()
model_dogrusal.fit(x_ölçekli,y_train)
y_dogrusal_predict= model_dogrusal.predict(x_ölçekli_test)
mse_dogrusal= mean_squared_error(y_test,y_dogrusal_predict)
r2_dogrusal = r2_score(y_test, y_dogrusal_predict)
```

# Ridge Regresyon

```python
ridge= Ridge(alpha=1.0)
ridge.fit(x_ölçekli,y_train)
y_ridge_predict= ridge.predict(x_ölçekli_test)
mse_ridge=mean_squared_error(y_test,y_ridge_predict)
r2_ridge = r2_score(y_test, y_ridge_predict)
```

# PCR

```python
pca=PCA(n_components=5)
x_pca=pca.fit_transform(x_ölçekli)
xx_pca=pca.transform(x_ölçekli_test)
model_pcr=LinearRegression()
model_pcr.fit(x_pca,y_train)
y_pcr_predict=model_pcr.predict(xx_pca)
mse_pcr=mean_squared_error(y_test,y_pcr_predict)
r2_pcr = r2_score(y_test, y_pcr_predict)
```

# PLS

```python
pls= PLSRegression(n_components=5)
pls.fit(x_ölçekli,y_train)
y_pls_predict=pls.predict(x_ölçekli_test)
mse_pls= mean_squared_error(y_test,y_pls_predict)
r2_pls = r2_score(y_test, y_pls_predict)
```

# tablolama

```python
isimler = [
    "Doğrusal Regresyon",
    "Ridge (α=1.0)",
    "PCR (5 bileşen)",
    "PLS (5 bileşen)"
]
mse = [
    mse_dogrusal,
    mse_ridge,
    mse_pcr,
    mse_pls
]

r2 = [
    r2_dogrusal,
    r2_ridge,
    r2_pcr,
    r2_pls
]
tablo = pd.DataFrame({
    "Model": isimler,
    "MSE": mse,
    "R2": r2
})
print(tablo)
```
