```python
kitap = []
xxx=["Kürk Mantolu Madonna;Sabahattin Ali",
"İçimizdeki Şeytan;Sabahattin Ali",
"Tutunamayanlar;Oğuz Atay",
"Çalıkuşu;Reşat Nuri Güntekin",
"Saatleri Ayarlama Enstitüsü;Ahmet Hamdi Tanpınar",
"Aylak Adam;Yusuf Atılgan",
"Suç ve Ceza;Fyodor Dostoyevski",
"Karamazov Kardeşler;Fyodor Dostoyevski"]

try:
    kutuphane = open("kutuphane.txt", "r", encoding="utf-8")
    satirlar = kutuphane.readlines()
    for satir in satirlar:
        satir = satir.strip()
        if satir:
            bilgiler = satir.split(";")
            kitap.append(bilgiler)
    kutuphane.close()
except FileNotFoundError:
    kutuphane = open("kutuphane.txt", "w", encoding="utf-8")

    kutuphane.close()

print("Merhabalar kütüphanemize hoşgelniz.")
islem_devam=True
while islem_devam:
  siran_numarasi=1
  islem= int(input("""
      Lütfen yapmak istediğiniz işlem için ilgili numarayı tuşlayınız:
      ● 1 - Kitap Ekle/Kitap İadesi
      ● 2 - Kitap Çıkar/ Kitap Ödünç al
      ● 3 - Kitapları Listele
      ● 4 - Çıkış
  """))

  if islem==1:
    kutuphane = open("kutuphane.txt", "a", encoding="utf-8")
    eklenecek_kitapadi=input("Eserin adını giriniz: ")
    eklenecek_yazar=input("İade ediceğiniz/ Ekliyeceğiniz kitabın yazarını giriniz: ")

    eklenecek= [eklenecek_kitapadi ,eklenecek_yazar]
    if eklenecek in kitap:
      print("Bu kitap zaten kütüphanede kayıtlı.")
    else:
      kitap.append(eklenecek)
      dosyayaekle= eklenecek_kitapadi + ";"+ eklenecek_yazar
      kutuphane.write("\n"+dosyayaekle)
      print(eklenecek_kitapadi,"adlı kitabınız başarıyla kütüphaneye eklenmiştir")
      kutuphane.close()
    devi= input("Eğer işleminizi bitirmek istiyorsanız lütfen -B- harfini tuşlayınız. Devam etmek için ise rastgele bir tuşa basınız.")
    if devi=="B" or devi=="b":
          islem_devam=False
          print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")





  elif islem==2:
    a=1
    b=0
    count=0


    yazarınkitabı=[]
    aradıgıyazar= input("Kitabın yazarını giriniz: ")

    try:

      for all in kitap:
              string=all[1]
              stringg=string.lower()

              if aradıgıyazar.lower() == stringg:
                yazarınkitabı.append(all)
                print(a, yazarınkitabı[b][0])
                a+=1
                b+=1
                count+=1
      if count==0:
        print("Aradığınız yazarın kitabı sistemimizde bulunmmaktadır.")
      else:

        kitapno= int(input("Kaç numaralı kitabı istiyorsunuz: "))
        if  kitapno < 1 or kitapno > count:
                    print("Sanırım yanlış sayıyı tuşladınız. Lütfen tekrar deneyiniz.")
        else:
                    print(yazarınkitabı[kitapno-1][0], "adlı kitabı aldınız. İyi okumalar dileriz.")
                    ddd = yazarınkitabı[kitapno-1]
                    kitap.remove(ddd)
                    kutuphane = open("kutuphane.txt", "w", encoding="utf-8")
                    for sıkıldım in kitap:
                       kutuphane.write(sıkıldım[0] + ";" + sıkıldım[1]+"\n")
                    kutuphane.close()



    except ValueError:
              print("Lütfen sayı girip tekrar deneyiniz")










    devi= input("Eğer işleminizi bitirmek istiyorsanız lütfen -B- harfini tuşlayınız. Devam etmek için ise rastgele bir tuşa basınız.")
    if devi=="B" or devi=="b":
          islem_devam=False
          print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")




  elif  islem==3:
    if kitap:

      for kitapp in kitap:
          print(siran_numarasi, "-", kitapp[0], "-", kitapp[1])
          siran_numarasi+=1
    else:
      print("Maalesef şu an kütüphanemizde mevcut bir kitap bulunmamaktadır. Lütfen daha sonra tekrar deneyiniz.")
      print(".\n.\n.\n.\nAnlık olarak kütüphanemize kitap iadeleri yapıldı isterseniz başa dönüp işlemi tekrar gerçekleştirebilirsiniz.")
      kutuphane = open("kutuphane.txt", "w", encoding="utf-8")
      for lll in xxx:
        kitap.append(lll.split(";"))
        kutuphane.write(lll+"\n")
      kutuphane.close()


    devammi= input("Eğer işleminizi bitirmek istiyorsanız lütfen -B- harfini tuşlayınız. Devam etmek için ise rastgele bir tuşa basınız.")
    if devammi=="B" or devammi=="b":
          islem_devam=False
          print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")











  elif islem==4:
    print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")
    islem_devam= False

  else:
    print("Sanırım yanlış bir numara tuşladınız. Lütfen daha dikkat ederek tekrardan tuşlayınız." ,"\nSistem yeniden başlatılıyor...")
    islem_devam= True

```
