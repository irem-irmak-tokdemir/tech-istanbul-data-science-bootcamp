```python
from google.colab import drive
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,Ridge
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
drive.mount("/content/drive")
from sklearn.decomposition import PCA
from sklearn.cross_decomposition import PLSRegression
from sklearn.neural_network import MLPRegressor
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import fashion_mnist
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, ReLU
from keras.models import Sequential
from sklearn.metrics import accuracy_score
import numpy as np
```

```python
print("Merhabalar uygulamamıza hoşgeldiniz. \nYapacağınız işlemi seçmek için işlemin sıra numarasını tuşlayınız")
whiledöngüsü=True
while(whiledöngüsü):
  işlem_no=input("""
        1 - Regresyon Analizi
        2 - Kümeleme Analizi
        3 - Metin Duygu Analizi
        4 - Görüntü Sınıflandırma
        5 - Çıkış
        """)
  if işlem_no=="1":

      veri_yolu_evfiyatı="/content/drive/My Drive/Tech İstanbul/proje_ev_fiyatlari.csv"
      ev_fiyatı=pd.read_csv(veri_yolu_evfiyatı)

      x_ev=ev_fiyatı[["alan", "oda_sayisi", "bina_yasi","merkeze_uzaklik", "kat"]]
      y_ev=ev_fiyatı[["fiyat"]]
      x_train_ev,x_test_ev,y_train_ev,y_test_ev= train_test_split(x_ev,y_ev,test_size=0.2,random_state=42)

      ölçek=StandardScaler()
      x_train_ev_ölçekli=ölçek.fit_transform(x_train_ev)
      x_test_ev_ölçekli=ölçek.transform(x_test_ev)
      ölçeky=StandardScaler()
      y_train_ev_ölçekli=ölçeky.fit_transform(y_train_ev)
      y_test_ev_ölçekli=ölçeky.transform(y_test_ev)

      #Doğrusal regresyon

      model_dogrusal=LinearRegression()
      model_dogrusal.fit(x_train_ev_ölçekli,y_train_ev)
      y_tahmin_ev=model_dogrusal.predict(x_test_ev_ölçekli)
      mse_dogrusal=mean_squared_error(y_test_ev,y_tahmin_ev)
      r2_dogrusal=r2_score(y_test_ev,y_tahmin_ev)


      # ridge
      ridge_ev=Ridge(alpha=0.994)
      ridge_ev.fit(x_train_ev_ölçekli,y_train_ev)
      y_predict_ridge=ridge_ev.predict(x_test_ev_ölçekli)
      mse_ridge=mean_squared_error(y_test_ev,y_predict_ridge)
      r2_ridge=r2_score(y_test_ev,y_predict_ridge)



      # pcr
      pca=PCA(n_components=3)
      x_pca=pca.fit_transform(x_train_ev_ölçekli)
      x_pca_test=pca.transform(x_test_ev_ölçekli)
      model_pcr= LinearRegression()
      model_pcr.fit(x_pca,y_train_ev)
      y_pred_pca=model_pcr.predict(x_pca_test)
      mse_pcr=mean_squared_error(y_test_ev,y_pred_pca)
      r2_pcr=r2_score(y_test_ev,y_pred_pca)


      # pls
      pls=PLSRegression(n_components=3)
      pls.fit(x_train_ev_ölçekli,y_train_ev)
      y_pls_predict=pls.predict(x_test_ev_ölçekli)
      mse_pls=mean_squared_error(y_test_ev,y_pls_predict)
      r2_pls=r2_score(y_test_ev,y_pls_predict)


      #yapay sinir ağı

      model_ysa=MLPRegressor(max_iter=1000,random_state=22)#2,14,88,38,22
      model_ysa.fit(x_train_ev_ölçekli, y_train_ev_ölçekli)
      y_ysa_predict=model_ysa.predict(x_test_ev_ölçekli)
      y_ysa_predist_real=y_ysa_predict*ölçeky.scale_[0]+ölçeky.mean_[0]
      x_tum_scaled=ölçek.transform(x_ev)
      y_ysa_tum=model_ysa.predict(x_tum_scaled)
      y_yapay_tum_gerçek=y_ysa_tum*ölçeky.scale_[0]+ölçeky.mean_[0]
      mse_ysa=mean_squared_error(y_test_ev,y_ysa_predist_real)
      r2_ysa=r2_score(y_test_ev,y_ysa_predist_real)


      isimler = [
          "---------------------",
          "Doğrusal Regresyon",
          "Ridge",
          "PCR",
          "PLS",
          "Yapay Sinir Ağı"

      ]
      mse = [
          "---------------------",
          f"{mse_dogrusal: .0f}",
          f"{mse_ridge:.0f}",
          f"{mse_pcr:.0f}",
          f"{mse_pls:.0f}",
          f"{mse_ysa:.0f}"
      ]

      r2 = [
          "---------------------",
          f"{r2_dogrusal:.3f}",
          f"{r2_ridge:.3f}",
          f"{r2_pcr:.3f}",
          f"{r2_pls:.3f}",
          f"{r2_ysa:.3f}"
      ]
      tablo = pd.DataFrame({
          "Model": isimler,
          "MSE": mse,
          "R^2": r2
      })
      print(tablo.to_string(index=False))
      sonuçlar = open("sonuclar.txt", "a", encoding="utf-8")

      stringtablo=str(tablo.to_string(index=False))
      sonuçlar.write("\n"+stringtablo)
      sonuçlar.close()
  elif işlem_no=="2":
      veri_yolu_müşteri="/content/drive/My Drive/Tech İstanbul/proje_musteri_segmentasyonu.csv"
      müşteri=pd.read_csv(veri_yolu_müşteri)
      x=müşteri[["yillik_gelir","harcama_puani"]]
      ölçekk=StandardScaler()
      X=ölçekk.fit_transform(x)

      inertia = []
      for centerr in range(2,16):
        kmeans = KMeans(n_clusters=centerr, random_state=42)
        kmeans.fit(X)
        inertia.append(kmeans.inertia_)
      plt.plot(range(2,16), inertia, marker="o")
      plt.xlabel("Küme Sayısı")
      plt.ylabel("Inertia")
      plt.title("Elbow Yöntemi")
      plt.show()

      kmeans=KMeans(n_clusters=5)
      kmeans.fit(X)
      centers=kmeans.cluster_centers_
      labels=kmeans.labels_



      plt.figure(figsize=(10,6))
      plt.scatter(X[:,0],X[:,1],s=50,c=labels,label="Müşteri")
      plt.scatter(centers[:,0],centers[:,1],c="red",s=200,alpha=0.9,label="Küme Merkezi",marker="X")
      plt.title("Müşteriler")
      plt.xlabel("Yıllık Gelir")
      plt.ylabel("Harcama Puanı")
      plt.show()

      silhouette=silhouette_score(X,labels)

      sonuc_kümleme = f"Silhouette skoru: {silhouette:.2f}"
      sonuçlar = open("sonuclar.txt", "a", encoding="utf-8")
      sonuçlar.write("\nKümeleme Analizi"+"\n"+sonuc_kümleme)
      sonuçlar.close()

      print("Silhouette skoru:",f"{silhouette: .2f}")
  elif işlem_no=="3":
      import kagglehub
      path = kagglehub.dataset_download("lakshmi25npathi/imdb-dataset-of-50k-movie-reviews")
      import os
      os.listdir(path)
      veri=pd.read_csv(os.path.join(path, "IMDB Dataset.csv"))
      veri.loc[veri["sentiment"]=="positive","sentiment"]=1
      veri.loc[veri["sentiment"]=="negative","sentiment"]=0
      X_m=veri[["review"]]
      y_m=veri["sentiment"].astype(int)

      X_train_metin,X_test_metin,y_train_metin,y_test_metin=train_test_split(
          X_m,y_m,test_size=0.2,random_state=42
      )
      tokenizer = Tokenizer(num_words=10000)


      tokenizer.fit_on_texts(X_train_metin["review"])


      sayisal_karsilik = tokenizer.texts_to_sequences(X_train_metin["review"])
      sabit_uzunluk = pad_sequences(sayisal_karsilik, maxlen=200)


      sayisal_karsilikk = tokenizer.texts_to_sequences(X_test_metin["review"])
      sabit_uzunlukk = pad_sequences(sayisal_karsilikk, maxlen=200)

      from keras.layers import Embedding
      from tensorflow.keras.models import Sequential
      from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

      model = Sequential([
          Embedding(input_dim=10000, output_dim=32),
          SimpleRNN(32),
          Dense(1, activation="sigmoid"),
      ])

      model.compile(
          loss="binary_crossentropy",
          optimizer="adam",
          metrics=["accuracy"]
      )

      model.fit(
          sabit_uzunluk,
          y_train_metin,
          epochs=5,
          batch_size=64,
          validation_split=0.2
      )

      kayip, dogruluk = model.evaluate(
          sabit_uzunlukk,
          y_test_metin
      )

      print("\nTest doğruluğu:", dogruluk)

      sonuçlar = open("sonuclar.txt", "a", encoding="utf-8")
      mm="\nMetin Duygu Analizi"+"\nTest doğruluğu:"+ f"{dogruluk:.2f}"
      sonuçlar.write(mm)
      sonuçlar.close()
  elif işlem_no==4:
      (X_train,y_train),(X_test,y_test)=fashion_mnist.load_data()
      X_train=X_train.reshape(-1,28,28,1)/255
      X_test=X_test.reshape(-1,28,28,1)/255
      cnn_model=Sequential([
          Conv2D(32,(3,3),activation='relu',input_shape=(28,28,1)),
          MaxPooling2D((2,2)),
          Conv2D(64,(3,3)),
          ReLU(),
          MaxPooling2D((2,2)),
          Conv2D(128,(3,3)),
          ReLU(),
          MaxPooling2D((2,2)),
          Flatten(),
          Dense(128,activation='relu'),
          Dense(64,activation='relu'),
          Dense(32,activation='relu'),
          Dense(10,activation='softmax'),


      ])
      cnn_model.compile(
          optimizer='adam',
          loss='sparse_categorical_crossentropy',
          metrics=['accuracy']
      )

      cnn_model.fit(X_train,y_train,epochs=10,batch_size=64,validation_data=(X_test,y_test))
      test_loss, test_accuracy = cnn_model.evaluate(X_test, y_test)


      predict = np.argmax(cnn_model.predict(X_test), axis=1)
      basari= (accuracy_score(predict,y_test))
      print("Test doğruluğu:", test_accuracy)
      print("Başarı skoru:",basari)

      sonuc = f"Test doğruluğu: {test_accuracy}"
      sonucc = f"Başarı skoru: {basari}"

      sonuçlar = open("sonuclar.txt", "a", encoding="utf-8")
      sonuçlar.write("\nGörüntü sınıflandırma"+"\n"+sonuc+"\n"+sonucc)
      sonuçlar.close()
  elif işlem_no=="5":
    print("Uygulamadan çıkılıyor.")
    whiledöngüsü=False
    break
  else:
    print("Yanlış bir sayı tuşlamış gibi görünüyorsunuz.")


  sorusorma=True
  while(sorusorma):
     davammı=input("İşlemlerinize devam etmek ister misiniz? \nEvet için 1'i hayır için 2'yi tuşlayınız")
     if davammı=="1":
       whiledöngüsü=True
       sorusorma=False
     elif davammı=="2":
       print("Uygulamadan çıkılıyor.")
       whiledöngüsü=False
       sorusorma=False
     else:
       print("Yanlış bir sayı tuşlamış gibi görünüyorsunuz.")
```
