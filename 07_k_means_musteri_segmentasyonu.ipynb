```python
from google.colab import drive
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
drive.mount("/content/drive")
veri_yolu="/content/drive/MyDrive/Tech İstanbul/musteri_segmentasyonu.csv"
veri=pd.read_csv(veri_yolu)
```

```python
X = veri[["yillik_gelir", "harcama_puani"]]
ölçek= StandardScaler()
X_ölcekli=ölçek.fit_transform(X)
```

```python
k_degerleri = range(1, 11)      # 1'den 10'a kadar K
inertia_listesi = []
for k in k_degerleri:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_ölcekli)      # olceklenmis veri
    inertia_listesi.append(kmeans.inertia_)
# Farkli K degerleri icin dirsek (elbow) grafigi
plt.plot(k_degerleri, inertia_listesi, marker="o")
plt.xlabel("Kume sayisi (K)")
plt.ylabel("Inertia (WCSS)")
plt.show()
```

```python
from numpy._core.defchararray import center
kmeans=KMeans(n_clusters=5)
kmeans.fit(X_olcekli)
Centers=kmeans.cluster_centers_
label=kmeans.labels_
plt.figure(figsize=(12,8))
renkler = ["yellow", "orange", "magenta", "cyan", "pink"]
for i in range(5):
    plt.scatter(
        X_ölçekli[label == i,0],
        X_ölçekli[label == i,1],
        s=50,
        c=renkler[i],

        label=f"Küme {i+1}"
    )
plt.scatter(Centers[:,0],Centers[:,1],c="black",s=200,alpha=0.99,label="Küme merkezleri",marker="X")

plt.title("Müşteriler")
plt.xlabel("Yıllık Gelir")
plt.ylabel("Harcama puanı")
plt.legend()
plt.show()
print("""
    Grafikteki Yeri         Yıllık Kazanç          Harcama Puanı
      Sol-Üst                 Düşük                  Yüksek
      Sol-Alt                 Düşük                  Düşük
      Sağ-Üst                 Yüksek                 Yüksek
      Sağ-Alt                 Yüksek                 Düşük
       Orta                   Ortalama               Ortalama


""")
```
