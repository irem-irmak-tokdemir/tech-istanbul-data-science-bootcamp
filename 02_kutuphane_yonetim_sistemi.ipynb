```python
islem_devam=True
kitap_list=["Tehlikeli Oyunlar — Oğuz Atay","Korkuyu Beklerken — Oğuz Atay","Tutunamayanlar - Oğuz Atay",
              "Suç ve Ceza — Fyodor Dostoyevski","1984 — George Orwell","Beyaz Diş — Jack London","Don Kişot — Miguel de Cervantes","Hayvan Çiftliği — George Orwell",
              "Vahşetin Çağrısı — Jack London"]
kitap_sözlük=kitap_sözlük = {
    "Oğuz Atay": [
        "Tehlikeli Oyunlar",
        "Korkuyu Beklerken",
        "Tutunamayanlar"
    ],

    "Fyodor Dostoyevski": [
        "Suç ve Ceza"
    ],

    "George Orwell": [
        "1984",
        "Hayvan Çiftliği"
    ],

    "Jack London": [
        "Beyaz Diş",
        "Vahşetin Çağrısı"
    ],

    "Miguel de Cervantes": [
        "Don Kişot"
    ]
}
print("Merhabalar kütüphanemize hoşgelniz.")

while islem_devam:
  siran_numarasi=1
  islem= int(input("""
      Lütfen yapmak istediğiniz işlem için ilgili numarayı tuşlayınız:
      ● 1 - Kitap Ekle/Kitap İadesi
      ● 2 - Kitap Çıkar/ Kitap Ödünç al
      ● 3 - Kitapları Listele
      ● 4 - Çıkış
  """))

  if islem==1:
    eklenecek_yazar=input("İade ediceğiniz/ Ekliyeceğiniz kitabın yazarını giriniz: ")
    eklenecek_kitapadi=input("Eserin adını giriniz: ")
    eklenecek= eklenecek_kitapadi+" - "+eklenecek_yazar


    if eklenecek_yazar in kitap_sözlük:
     kitap_sözlük[eklenecek_yazar].append(eklenecek_kitapadi)
    else:

     kitap_sözlük[eklenecek_yazar] = [eklenecek_kitapadi]

    if eklenecek not in kitap_list:
      kitap_list.append(eklenecek)
      print("Kitabınız kütüphaneye başarıyla eklendi.")

    else:
      print("Yazdığınız kitap kütüphanede zaten  mevcut.")
    devami= input("Eğer işleminizi bitirmek istiyorsanız lütfen -B- harfini tuşlayınız. Devam etmek için ise rastgele bir tuşa basınız.")
    if devami=="B" or devami=="b":
      islem_devam=False
      print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")




  elif islem==2:
    aranankitap=[]
    yazar_bulundu_mu= False
    istenen=input("Hangi yazarın kitabını almak istersiniz: ")
    a=1
    for yazarcık in kitap_sözlük:
      if istenen.lower()==yazarcık.lower():
        yazar_bulundu_mu=True
        for kitap in kitap_sözlük[yazarcık]:
            aranankitap.append(kitap)

    if yazar_bulundu_mu==False:
      print("Aradığınız yazarın kitabı bulunmamaktadır.")

    elif yazar_bulundu_mu== True:
      for kitapcık in aranankitap:
        print(a, ".",kitapcık )
        a+=1
      istenenkitapno=int(input("Kaç numaralı kitabı istiyorsunuz: "))

      if istenenkitapno > len(aranankitap) or istenenkitapno < 1:
        print("Geçersiz seçim.")
      else:
        print(aranankitap[istenenkitapno-1], "adlı kitabı aldınız. İyi okumalar dilerim.")



        for kitaap in kitap_list:
          if aranankitap[istenenkitapno-1] in kitaap:
             kitap_list.remove(kitaap)

        for yazarcık in kitap_sözlük:
          if istenen.lower() == yazarcık.lower():
              for kitap in kitap_sözlük[yazarcık]:
                   if kitap == aranankitap[istenenkitapno-1]:
                        kitap_sözlük[yazarcık].remove(kitap)







    devi= input("Eğer işleminizi bitirmek istiyorsanız lütfen -B- harfini tuşlayınız. Devam etmek için ise rastgele bir tuşa basınız.")
    if devi=="B" or devi=="b":
          islem_devam=False
          print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")




  elif  islem==3:
    if kitap_list:
      for kitap in kitap_list:
          print(siran_numarasi, "- ",kitap)
          siran_numarasi+=1
    else:
      print("Maalesef şu an kütüphanemizde mevcut bir kitao bulunmamaktadır. Lütfen daha sonra tekrar deneyiniz.")

    devammi= input("Eğer işleminizi bitirmek istiyorsanız lütfen -B- harfini tuşlayınız. Devam etmek için ise rastgele bir tuşa basınız.")
    if devammi=="B" or devammi=="b":
          islem_devam=False
          print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")











  elif islem==4:
    print("Bizi tercih ettiğiniz için teşekkür ederiz. Yine bekleriz.","\nSistem kapatılıyor...")
    islem_devam= False

  else:
    print("Sanırım yanlış bir numara tuşladınız. Lütfen daha dikkat ederek tekrardan tuşlayınız." ,"\nSistem yeniden başlatılıyor...")
    islem_devam= True
```
